```python

def plot_cosmo_contours(acm_chain, label, color, 
                        params_to_plot = None,#['omega_cdm', 'sigma8_m', 'n_s'],
                        truth_color=None,
                        chain_consumer = None, make_plot = True,
                        smooth = 50, bins = 10
                       ):

    if truth_color is None:
        truth_color = color

    samples = acm_chain['samples']

    weights = acm_chain['weights']

    ranges = acm_chain['ranges']

    names = np.array(acm_chain['names']) # for under the hood

    labels = acm_chain['labels'] # for plots

    markers = acm_chain['markers']

    if params_to_plot is None:
        params_to_plot = names

    select_paramters = np.array([np.where(names==param)[0][0] for param in params_to_plot])

    labels_to_plot = [labels[param] for param in params_to_plot]
    samples_frame = pd.DataFrame(data=samples[:,select_paramters], columns=labels_to_plot)

    extents = {}
    truths = {}
    
    for param in params_to_plot:
        
        extents[param] = ranges[param]
        truths[param] = markers[param]
    
    if chain_consumer is None:
        c = ChainConsumer()
    else:
        c = chain_consumer
    
    c.add_chain(Chain(samples=samples_frame, name=label, smooth=smooth, bins=bins, #3 and 30 or 50 and 10
                      color=mpl.colors.to_hex(color),
                      shade_alpha=0.8, linestyle=':'
                     )
               )
    c.set_plot_config(
        PlotConfig(
            extents={labels[k]: extents[k] for k in params_to_plot},
            label_font_size = 20,
            tick_font_size=12,
            summary_font_size=12,
            #contour_label_font_size = 15 # for legend?
        )
    )

    #print({k: truths[k] for k in labels_to_plot})
    c.add_truth(Truth(location={label: truths[param] for param, label in zip(params_to_plot, labels_to_plot)}, 
                      color=truth_color, 
                      edge_color = 'black',
                      alpha=0.5,
                      marker="X",
                      marker_edge_width=2
                     ))
    
    
    c.plotter.config.sigma2d=True

    if make_plot:
        
        fig = c.plotter.plot()
        return fig, c
        
    return c
```
